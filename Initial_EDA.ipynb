```python

#Perfil del DF 
def perfil(df, cols):
    return pd.DataFrame({
        "tipo": df[cols].dtypes.astype(str),
        "nulos_%": (df[cols].isna().mean() * 100).round(1),
        "unicos": df[cols].nunique(),
        "top_5": [df[c].value_counts(dropna=False).head(5).to_dict() for c in cols],
    })

#Normalizar textos 
def norm_txt(s):
    """quita espacios, pasa a minúsculas y convierte los falsos nulos en NA"""
    s = s.astype("string").str.strip().str.lower()
    return s.replace({"": pd.NA, "n/a": pd.NA, "na": pd.NA, "null": pd.NA, "-": pd.NA, "none": pd.NA})

#Mapeo
def mapear(s, mapa, nombre):
    """aplica el mapeo y falla si aparece un valor no contemplado"""
    s = norm_txt(s)
    sin_mapear = set(s.dropna().unique()) - set(mapa)
    assert not sin_mapear, f"{nombre}: valores sin mapear {sin_mapear}"
    return s.map(mapa)
```

## Normalización de categóricas

**Columnas revisadas:** Sistema Operativo, Versión SO, Estado TPM, Encryption Readiness,
Estado de Cifrado, Tecnología de cifrado State, Estado de Cumplimiento, Fuente de Datos,
En MDM, En AD, En MacOS

**Prueba:** se compararon los valores únicos de cada columna antes y después de normalizar
el texto (sin espacios extra, sin mayúsculas, sin acentos).

**Resultado:** el número de valores únicos no cambió en ninguna columna. No existen variantes
de escritura de una misma categoría.

**Decisión:** no se aplica mapeo de categorías. Los valores se conservan como vienen en la fuente.

**Nota para el pipeline:** la validación de valores permitidos se mantiene, para detectar
categorías nuevas cuando el cliente suba un Excel actualizado.

```python
# Estandarizacion de columnas (snake case)

import unicodedata
import janitor

def sin_acentos(texto):
    """Versión → Version, Tecnología → Tecnologia"""
    return "".join(ch for ch in unicodedata.normalize("NFKD", str(texto))
                   if not unicodedata.combining(ch))

def limpiar_nombres(df):
    """snake_case, sin acentos ni caracteres especiales, separa CamelCase (EquipoID → equipo_id)"""
    out = df.rename(columns=sin_acentos)   # quitar acentos ANTES de janitor
    out = out.clean_names(
        case_type="snake",
        remove_special=True,
        strip_underscores=True,
    )
    assert out.columns.is_unique, "dos columnas quedaron con el mismo nombre"
    return out

df_a = limpiar_nombres(df_a)
cat_cols = [c for c in df_a.columns if c != "equipo_id"]
df_a.columns.tolist()
```

### 2. Estandarización de nombres de columnas

Los nombres originales mezclan mayúsculas, acentos, espacios, CamelCase y palabras en
inglés y español (ej. `EquipoID`, `Versión SO`, `Tecnología de cifrado State`). Esto complica
su uso en código, en SQL y en el backend.

Se aplicó la función `limpiar_nombres()` (basada en `pyjanitor.clean_names`), que:

- convierte todo a **snake_case** (`EquipoID` → `equipo_id`)
- elimina **acentos** (`Versión SO` → `version_so`)
- elimina **espacios y caracteres especiales**
- valida que no queden **nombres duplicados**


```python
#Revision de llave (equipo_id)

# formato
antes = df_a["equipo_id"].nunique()
df_a["equipo_id"] = df_a["equipo_id"].astype("string").str.strip().str.upper()
print("únicos antes/después de limpiar formato:", antes, df_a["equipo_id"].nunique())

# nulos
print("nulos:", df_a["equipo_id"].isna().sum())

# duplicados
dup_mask = df_a.duplicated("equipo_id", keep=False)
print("filas con id repetido:", dup_mask.sum())
print("total filas:", len(df_a), "| ids únicos:", df_a["equipo_id"].nunique())
print("filas 100% idénticas:", df_a.duplicated(keep=False).sum())
```

### 3. Validación de la llave `equipo_id`

Se verificó que `equipo_id` identifique de forma única a cada equipo:

- **Formato:** se eliminaron espacios y se unificaron mayúsculas; el número de IDs únicos no cambió.
- **Nulos:** 0 registros sin ID.
- **Duplicados:** 0 IDs repetidos; el número de filas coincide con el de IDs únicos.

**Conclusión:** cada fila representa exactamente un equipo. `equipo_id` es una llave válida
para unir con el dataset de Persona B y para los conteos del dashboard. No se eliminó ninguna fila.

```python
# 4.4 Agrupar version_so en familia de SO (la columna original se conserva)

ETIQUETA_SIN_DATO = "No disponible"

def familia_so(v):
    v = str(v)
    if v == ETIQUETA_SIN_DATO or v.lower() in ("nan", "unknown", ""):
        return ETIQUETA_SIN_DATO
    if v.startswith("10.0.19") or v == "Windows 10":
        return "Windows 10"
    if v.startswith(("10.0.22", "10.0.26")) or v == "Windows 11":
        return "Windows 11"
    if v.startswith("Windows"):
        return "Windows (sin versión)"
    if v.startswith(("macOS", "26.")):
        return "macOS"
    return "Sin clasificar"          # versión nueva no contemplada → revisar

df_a["familia_so"] = df_a["version_so"].map(familia_so)
cat_cols.append("familia_so")
print("sin clasificar:", (df_a["familia_so"] == "Sin clasificar").sum())
df_a["familia_so"].value_counts()
```

```python
# Perfilado de valores faltantes

import re

PATRON_SIN_DATO = r"^\s*(no disponible|desconocido|sin dato|sin informacion|n/?a|na|null|none|unknown|-)\s*$"

def es_sin_dato(s):
    """True si el valor es texto tipo 'No disponible', 'N/A', etc."""
    return s.astype("string").map(lambda x: sin_acentos(x).lower() if pd.notna(x) else x) \
            .str.match(PATRON_SIN_DATO, na=False)

faltantes = pd.DataFrame({
    "nulos_reales": df_a[cat_cols].isna().sum(),
    "nulos_disfrazados": pd.Series({c: es_sin_dato(df_a[c]).sum() for c in cat_cols}),
})
faltantes["total"] = faltantes.sum(axis=1)
faltantes["%"] = (faltantes["total"] / len(df_a) * 100).round(1)
faltantes.sort_values("%", ascending=False)
```

```python
# 9.1 % de cifrado por cada dimensión
COLORES_CIFRADO = {"Encrypted": AZUL, "Not encrypted": "#C0392B", ETIQUETA_SIN_DATO: GRIS}

def cifrado_por(dim, ax):
    t = pd.crosstab(df_a[dim], df_a["estado_de_cifrado"], normalize="index").mul(100)
    t = t.reindex(columns=[c for c in COLORES_CIFRADO if c in t.columns]).sort_values("Encrypted")
    izq = np.zeros(len(t))
    for estado in t.columns:
        ax.barh(t.index.astype(str), t[estado], left=izq, color=COLORES_CIFRADO[estado],
                height=0.6, edgecolor="white", linewidth=1, label=estado)
        izq += t[estado].values
    n = df_a[dim].value_counts().reindex(t.index)
    for i, (k, v) in enumerate(n.items()):
        nc = t.loc[k].get("Not encrypted", 0)
        ax.text(101, i, f"n={v:,} · no cifr. {nc:.1f}%", va="center", fontsize=8, color="#666")
    ax.set_xlim(0, 135)
    ax.set_title(dim, loc="left", fontsize=11, fontweight="bold", color=TINTA)
    ax.spines[["top", "right"]].set_visible(False)

dims = ["sistema_operativo", "familia_so", "version_so", "estado_tpm", "encryption_readiness",
        "tecnologia_de_cifrado_state", "fuente_de_datos", "en_mdm", "en_ad"]
nrows = int(np.ceil(len(dims) / 2))
fig, axes = plt.subplots(nrows, 2, figsize=(15, 3.2 * nrows))
for ax, dim in zip(axes.flat, dims):
    cifrado_por(dim, ax)
for ax in axes.flat[len(dims):]:
    ax.axis("off")
handles, labels = axes.flat[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper right", ncol=3, frameon=False)
fig.suptitle("Estado de cifrado por dimensión (% de equipos)", x=0.01, ha="left", fontsize=13, fontweight="bold")
plt.tight_layout(rect=(0, 0, 1, 0.97))
plt.show()
```

Se aplicó la misma `limpiar_nombres()` de la parte A, así que los nombres siguen la misma convención
(`País` → `pais`, `Mes Último Logon` → `mes_ultimo_logon`). Se validó que:

- la única columna en común con `df_a` es `equipo_id`;
- `equipo_id` tiene el mismo formato que en A, sin nulos ni duplicados;
- los IDs de A y B son exactamente los mismos, así que la unión es 1 a 1.

```python
# B.3 usuario_id: nulos y equipos por usuario

ETIQUETA_SIN_USUARIO = "No asignado"

u = df_b["usuario_id"].astype("string").str.strip()
formato_ok = u.str.match(r"^U-\d+$", na=False)
print("con formato U-####:", formato_ok.sum())
print("otros valores:", u[~formato_ok].value_counts(dropna=False).to_dict())

# unificar cualquier vacío o etiqueta de faltante como "No asignado"
sin_usuario = (u.isna() | es_sin_dato(u)
               | u.map(lambda x: sin_acentos(x).lower() if pd.notna(x) else x).eq("no asignado"))
df_b["usuario_id"] = u.mask(sin_usuario, ETIQUETA_SIN_USUARIO)

# flags
df_b["flag_sin_usuario"] = df_b["usuario_id"] == ETIQUETA_SIN_USUARIO
equipos_por_usuario_b = df_b.loc[~df_b["flag_sin_usuario"], "usuario_id"].value_counts()
df_b["n_equipos_usuario"] = (df_b["usuario_id"].map(equipos_por_usuario_b)
                             .fillna(0).astype("Int64"))

bitacora_b["usuario"] = {
    "equipos sin usuario": int(df_b["flag_sin_usuario"].sum()),
    "% sin usuario": round(df_b["flag_sin_usuario"].mean() * 100, 1),
    "usuarios distintos": int(equipos_por_usuario_b.size),
    "usuarios con más de 1 equipo": int((equipos_por_usuario_b > 1).sum()),
    "máximo de equipos de un usuario": int(equipos_por_usuario_b.max()) if len(equipos_por_usuario_b) else 0,
}
for k, v in bitacora_b["usuario"].items():
    print(f"{k}: {v}")

print("\nusuarios según cuántos equipos tienen:")
print(equipos_por_usuario_b.value_counts().sort_index().rename("usuarios").to_string())
```

### B.3 usuario_id

- `"No asignado"` no se trata como dato perdido sino como información: el equipo no tiene usuario.
  Se conserva la etiqueta y se agrega **`flag_sin_usuario`**.
- Se agrega **`n_equipos_usuario`** (cuántos equipos tiene el usuario de esa fila; 0 si no tiene).
- **Resultado:** _completar con los números de la celda anterior._

```python
# B.4 Normalización de categóricas (misma prueba que en la parte A)

CAT_B = ["pais", "region", "categoria_de_equipo", "clase_de_equipo",
         "perfil_de_privilegio", "estado_de_cuenta", "estado_hardware", "tipo_de_red"]

def limpiar_espacios(s):
    return s.astype("string").str.strip().str.replace(r"\s+", " ", regex=True)

def clave_txt(s):
    """sin espacios extra, sin mayúsculas, sin acentos"""
    return limpiar_espacios(s).map(lambda x: sin_acentos(x).lower() if pd.notna(x) else x)

prueba_b = pd.DataFrame({
    "unicos_antes": df_b[CAT_B].nunique(),
    "unicos_normalizado": pd.Series({c: clave_txt(df_b[c]).nunique() for c in CAT_B}),
})
prueba_b["variantes"] = prueba_b["unicos_antes"] - prueba_b["unicos_normalizado"]
prueba_b
```

```python
# Unificar variantes (solo cambia algo si la prueba encontró variantes) + unificar faltantes

cambios_b = {}
for c in CAT_B:
    s = limpiar_espacios(df_b[c])
    k = clave_txt(s)
    # cada variante se reemplaza por la forma más frecuente de esa categoría
    forma = (pd.DataFrame({"k": k, "v": s}).dropna()
               .groupby("k")["v"].agg(lambda g: g.value_counts().idxmax()))
    nueva = k.map(forma)
    cmb = {a: b for a, b in zip(s.dropna(), nueva.dropna()) if a != b}
    if cmb:
        cambios_b[c] = cmb
    # faltantes: mismo criterio que la parte A (5.2)
    mask = nueva.isna() | es_sin_dato(nueva)
    df_b[c] = nueva.mask(mask, ETIQUETA_SIN_DATO)

bitacora_b["variantes_unificadas"] = cambios_b
print("variantes unificadas:", cambios_b or "ninguna")
print()
print((df_b[CAT_B] == ETIQUETA_SIN_DATO).mean().mul(100).round(1)
      .sort_values(ascending=False).rename("% sin dato").to_string())
```

### B.4 Normalización de categóricas

**Columnas revisadas:** país, región, categoría, clase, perfil de privilegio, estado de cuenta,
estado de hardware, tipo de red.

**Prueba:** igual que en la parte A, se compararon los valores únicos antes y después de normalizar
el texto (sin espacios extra, sin mayúsculas, sin acentos).

**Resultado:** _completar con la tabla `prueba_b` (si `variantes` es 0 en todas, no había variantes)._

**Decisión:** las variantes, si las hay, se reemplazan por la forma más frecuente. Los faltantes
se unifican como **"No disponible"**, igual que en la parte A.

```python
# B.5 Coherencia de jerarquías (país → región, categoría ↔ clase)

def conflictos_jerarquia(df, padre, hijo):
    """valores de `padre` que aparecen con más de un `hijo` (sin contar 'No disponible')"""
    con_dato = df[(df[padre] != ETIQUETA_SIN_DATO) & (df[hijo] != ETIQUETA_SIN_DATO)]
    t = pd.crosstab(con_dato[padre], con_dato[hijo])
    return t[(t > 0).sum(axis=1) > 1]

# país → región
conf_pais = conflictos_jerarquia(df_b, "pais", "region")
print("países con más de una región:", conf_pais.index.tolist() or "ninguno")
display(conf_pais)

# región faltante: se completa con la región del país (solo países sin conflicto)
region_de_pais = (df_b[(df_b["pais"] != ETIQUETA_SIN_DATO) & (df_b["region"] != ETIQUETA_SIN_DATO)]
                  .groupby("pais")["region"].agg(lambda g: g.value_counts().idxmax())
                  .drop(conf_pais.index, errors="ignore"))
completar = (df_b["region"] == ETIQUETA_SIN_DATO) & df_b["pais"].isin(region_de_pais.index)
df_b["flag_region_imputada"] = completar
df_b.loc[completar, "region"] = df_b.loc[completar, "pais"].map(region_de_pais)
print("regiones completadas a partir del país:", completar.sum())
```

```python
# categoría ↔ clase: se revisa en los dos sentidos (solo se reporta, no se corrige)

conf_cat = conflictos_jerarquia(df_b, "categoria_de_equipo", "clase_de_equipo")
conf_clase = conflictos_jerarquia(df_b, "clase_de_equipo", "categoria_de_equipo")
print("categorías con más de una clase:", conf_cat.index.tolist() or "ninguna")
print("clases con más de una categoría:", conf_clase.index.tolist() or "ninguna")

bitacora_b["jerarquias"] = {
    "países con más de una región": conf_pais.index.tolist() or "ninguno",
    "regiones completadas a partir del país": int(completar.sum()),
    "categorías con más de una clase": conf_cat.index.tolist() or "ninguna",
    "clases con más de una categoría": conf_clase.index.tolist() or "ninguna",
}

pd.crosstab(df_b["categoria_de_equipo"], df_b["clase_de_equipo"], margins=True)
```

### B.5 Jerarquías

- **País → región:** cada país debería pertenecer a una sola región. Si un equipo no tiene región
  pero su país sí tiene una región clara, se completa y se marca con **`flag_region_imputada`**.
- **Categoría ↔ clase:** se revisa en los dos sentidos; solo se reporta, no se corrige.
- **Resultado:** _completar con lo que muestren las dos celdas anteriores._

```python
# B.6 Antigüedad: numérica + validar banda

def rango_banda(banda):
    """'0-12 meses' → (0, 12) · '37+ meses' → (37, inf) · '1-2 años' → (12, 24) · no entendible → (nan, nan)"""
    b = sin_acentos(str(banda)).lower()
    if banda is pd.NA or b in ("nan", "<na>", sin_acentos(ETIQUETA_SIN_DATO).lower()):
        return (np.nan, np.nan)
    factor = 12 if "ano" in b else 1
    if m := re.search(r"(\d+)\s*-\s*(\d+)", b):
        return (int(m[1]) * factor, int(m[2]) * factor)
    if m := re.search(r"(?:>|mas de)\s*(\d+)|(\d+)\s*\+", b):
        return (int(m[1] or m[2]) * factor, np.inf)
    if m := re.search(r"(?:<|menos de)\s*(\d+)", b):
        return (0, int(m[1]) * factor)
    return (np.nan, np.nan)

def orden_bandas(s):
    """bandas ordenadas por su inicio, con 'No disponible' al final"""
    vals = [v for v in s.dropna().unique() if v != ETIQUETA_SIN_DATO]
    vals = sorted(vals, key=lambda v: (np.isnan(rango_banda(v)[0]), rango_banda(v)[0]))
    return vals + ([ETIQUETA_SIN_DATO] if (s == ETIQUETA_SIN_DATO).any() else [])

def bandas_no_entendibles(s):
    return sorted({v for v in s.dropna().unique()
                   if v != ETIQUETA_SIN_DATO and np.isnan(rango_banda(v)[0])})


# banda: mismo tratamiento de faltantes que las categóricas
s = limpiar_espacios(df_b["banda_antiguedad"])
df_b["banda_antiguedad"] = s.mask(s.isna() | es_sin_dato(s), ETIQUETA_SIN_DATO)
ORDEN_BANDA_ANT = orden_bandas(df_b["banda_antiguedad"])

# meses: columna nueva numérica (la original se conserva)
original = df_b["antiguedad_meses"]
meses = pd.to_numeric(original.mask(es_sin_dato(original)), errors="coerce")
no_numericos = int((original.notna() & ~es_sin_dato(original) & meses.isna()).sum())
absurdos = (meses < 0) | (meses > 600)            # negativos o más de 50 años
df_b["antiguedad_meses_num"] = meses.mask(absurdos).astype("Float64")

# ¿los meses caen dentro de su banda?
rangos = df_b["banda_antiguedad"].map(rango_banda)
lo, hi = rangos.str[0], rangos.str[1]
validable = df_b["antiguedad_meses_num"].notna() & lo.notna()
dentro = (df_b["antiguedad_meses_num"].astype(float) >= lo) & (df_b["antiguedad_meses_num"].astype(float) <= hi)
df_b["flag_banda_antiguedad_incoherente"] = validable & ~dentro

bitacora_b["antiguedad"] = {
    "valores no numéricos": no_numericos,
    "valores absurdos (<0 o >600) → NaN": int(absurdos.sum()),
    "bandas no entendibles": bandas_no_entendibles(df_b["banda_antiguedad"]) or "ninguna",
    "filas validables (meses y banda)": int(validable.sum()),
    "banda incoherente con los meses": int(df_b["flag_banda_antiguedad_incoherente"].sum()),
    "con meses pero sin banda": int((df_b["antiguedad_meses_num"].notna() & (df_b["banda_antiguedad"] == ETIQUETA_SIN_DATO)).sum()),
    "con banda pero sin meses": int((df_b["antiguedad_meses_num"].isna() & (df_b["banda_antiguedad"] != ETIQUETA_SIN_DATO)).sum()),
}
for k, v in bitacora_b["antiguedad"].items():
    print(f"{k}: {v}")

(df_b.groupby("banda_antiguedad")["antiguedad_meses_num"]
     .agg(["count", "min", "max"]).reindex(ORDEN_BANDA_ANT))
```

### B.6 Antigüedad

- `antiguedad_meses` se conserva y se agrega **`antiguedad_meses_num`** (numérica; "No disponible",
  negativos y valores mayores a 600 meses quedan como NaN).
- Se comprueba que los meses caigan dentro del rango de su banda; las filas que no cumplen se marcan
  con **`flag_banda_antiguedad_incoherente`**.
- **Resultado:** _completar con la tabla min/max por banda y el conteo de incoherentes._

```python
# B.7 Último logon: fecha + validar banda de recencia

# Fecha de extracción de los datos. Si no se conoce, se usa el último día del mes más reciente.
FECHA_CORTE_B = None   # p. ej. pd.Timestamp("2025-11-30")

original = df_b["mes_ultimo_logon"]
if pd.api.types.is_datetime64_any_dtype(original):
    fecha = original
else:
    txt = limpiar_espacios(original)
    txt = txt.mask(es_sin_dato(txt))
    fecha = pd.to_datetime(txt, format="%Y-%m", errors="coerce")
    no_convertidas = int((txt.notna() & fecha.isna()).sum())
df_b["mes_ultimo_logon_fecha"] = fecha
no_convertidas = locals().get("no_convertidas", 0)

corte = FECHA_CORTE_B if FECHA_CORTE_B is not None else fecha.max() + pd.offsets.MonthEnd(0)

s = limpiar_espacios(df_b["banda_recencia"])
df_b["banda_recencia"] = s.mask(s.isna() | es_sin_dato(s), ETIQUETA_SIN_DATO)
ORDEN_BANDA_REC = orden_bandas(df_b["banda_recencia"])

# solo se sabe el mes, así que hay un rango de días posibles desde el último logon hasta el corte;
# la fila es incoherente solo si NINGÚN día de ese mes cae dentro de su banda
dias_min = (corte - (fecha + pd.offsets.MonthEnd(0))).dt.days.clip(lower=0)
dias_max = (corte - fecha).dt.days
rangos = df_b["banda_recencia"].map(rango_banda)
lo, hi = rangos.str[0], rangos.str[1]
validable = fecha.notna() & lo.notna()
df_b["flag_banda_recencia_incoherente"] = validable & ~((dias_max >= lo) & (dias_min <= hi))

bitacora_b["ultimo_logon"] = {
    "fecha de corte usada": str(pd.Timestamp(corte).date()) if pd.notna(corte) else None,
    "texto que no se pudo convertir": no_convertidas,
    "rango de fechas": f"{fecha.min().date()} a {fecha.max().date()}" if fecha.notna().any() else None,
    "fechas posteriores al corte": int((fecha > corte).sum()),
    "bandas no entendibles": bandas_no_entendibles(df_b["banda_recencia"]) or "ninguna",
    "filas validables (fecha y banda)": int(validable.sum()),
    "banda incoherente con la fecha": int(df_b["flag_banda_recencia_incoherente"].sum()),
}
for k, v in bitacora_b["ultimo_logon"].items():
    print(f"{k}: {v}")

(pd.crosstab(df_b["mes_ultimo_logon_fecha"].dt.strftime("%Y-%m"), df_b["banda_recencia"])
   .reindex(columns=[b for b in ORDEN_BANDA_REC if b != ETIQUETA_SIN_DATO]))
```

### B.7 Último logon

- Para validar `banda_recencia` hace falta la fecha de extracción (`FECHA_CORTE_B`); si no se conoce,
  se usa el fin del mes más reciente. Como solo hay mes, una fila se marca con
  **`flag_banda_recencia_incoherente`** solo si ningún día de ese mes cabe en su banda.
- **Resultado:** _completar con la tabla mes × banda y el conteo de incoherentes._

```python
# B.8.1 Distribución de cada columna categórica de B (mismo estilo que 8.2)

def barras_b(df, col, ax, orden=None):
    vc = df[col].value_counts()
    vc = vc.reindex([o for o in orden if o in vc.index])[::-1] if orden else vc.sort_values()
    pct = vc / len(df) * 100
    colores = [GRIS if v in (ETIQUETA_SIN_DATO, ETIQUETA_SIN_USUARIO) else AZUL for v in vc.index]
    ax.barh(vc.index.astype(str), vc.values, color=colores, height=0.6)
    for i, (k, n) in enumerate(vc.items()):
        ax.text(n, i, f"  {n:,} ({pct[k]:.1f}%)", va="center", fontsize=9, color=TINTA)
    ax.set_title(col, loc="left", fontsize=11, fontweight="bold", color=TINTA)
    ax.set_xlim(0, vc.max() * 1.35)
    ax.spines[["top", "right"]].set_visible(False)
    ax.tick_params(axis="x", colors="#999")

vista_b = df_b.assign(usuario=np.where(df_b["flag_sin_usuario"], ETIQUETA_SIN_USUARIO, "Con usuario"))
plots_b = [(c, None) for c in CAT_B] + [("banda_antiguedad", ORDEN_BANDA_ANT),
                                        ("banda_recencia", ORDEN_BANDA_REC), ("usuario", None)]
nrows = int(np.ceil(len(plots_b) / 2))
fig, axes = plt.subplots(nrows, 2, figsize=(14, 2.8 * nrows))
for ax, (col, orden) in zip(axes.flat, plots_b):
    barras_b(vista_b, col, ax, orden)
for ax in axes.flat[len(plots_b):]:
    ax.axis("off")
fig.suptitle("Parte B · distribución por columna (gris = No disponible / No asignado)",
             x=0.01, ha="left", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

### B.11 Unión A × B

`df_ab = df_a ⋈ df_b` por `equipo_id` (left join: base A, se añaden columnas de B).  
Se reporta cobertura de llaves (match, solo_en_a, solo_en_b) y se agrega `flag_sin_datos_b`  
para equipos sin par en B. Es la tabla principal para el dashboard.

```python
# B.11 Unión A × B por equipo_id

# ── 1. Cobertura de llaves ──────────────────────────────────────────────────
ids_a = set(df_a["equipo_id"].dropna())
ids_b = set(df_b["equipo_id"].dropna())

solo_en_a = ids_a - ids_b        # en A pero no en B
solo_en_b = ids_b - ids_a        # en B pero no en A
en_ambas  = ids_a & ids_b        # match exacto

print(f"Registros en A:        {len(df_a):>6,}")
print(f"Registros en B:        {len(df_b):>6,}")
print(f"IDs únicos en A:       {len(ids_a):>6,}")
print(f"IDs únicos en B:       {len(ids_b):>6,}")
print(f"\nMatch exacto (A∩B):    {len(en_ambas):>6,}  ({len(en_ambas)/len(ids_a)*100:.1f}% de A)")
print(f"Solo en A (sin par B): {len(solo_en_a):>6,}  → quedarán con NaN en cols de B")
print(f"Solo en B (sin par A): {len(solo_en_b):>6,}  → se descartan con left join")

# ── 2. Unión (left join: base = A, agrega cols de B cuando hay match) ───────
df_ab = df_a.merge(df_b, on="equipo_id", how="left", validate="one_to_one")

# flag para equipos de A sin par en B
df_ab["flag_sin_datos_b"] = df_ab["equipo_id"].isin(solo_en_a)

# ── 3. Resultado ─────────────────────────────────────────────────────────────
print(f"\ndf_ab shape:             {df_ab.shape}")
print(f"Equipos con datos de B:  {(~df_ab['flag_sin_datos_b']).sum():,}")
print(f"Equipos sin datos de B:  {df_ab['flag_sin_datos_b'].sum():,}")
assert len(df_ab) == len(df_a), "ERROR: la unión cambió el número de filas de A"
print("\nAssertion OK: df_ab conserva exactamente todas las filas de A")
df_ab.head()
```

```python
# % de cifrado por cada dimensión de B (mismo estilo que 9.1)

def cifrado_por_b(df, dim, ax, orden=None):
    t = pd.crosstab(df[dim], df["estado_de_cifrado"], normalize="index").mul(100)
    t = t.reindex(columns=[c for c in COLORES_CIFRADO if c in t.columns])
    t = t.reindex([o for o in orden if o in t.index][::-1]) if orden else t.sort_values("Encrypted")
    izq = np.zeros(len(t))
    for estado in t.columns:
        ax.barh(t.index.astype(str), t[estado], left=izq, color=COLORES_CIFRADO[estado],
                height=0.6, edgecolor="white", linewidth=1, label=estado)
        izq += t[estado].values
    n = df[dim].value_counts().reindex(t.index)
    for i, (k, v) in enumerate(n.items()):
        nc = t.loc[k].get("Not encrypted", 0)
        ax.text(101, i, f"n={v:,} · no cifr. {nc:.1f}%", va="center", fontsize=8, color="#666")
    ax.set_xlim(0, 135)
    ax.set_title(dim, loc="left", fontsize=11, fontweight="bold", color=TINTA)
    ax.spines[["top", "right"]].set_visible(False)

vista_ab = df_ab.assign(usuario=np.where(df_ab["flag_sin_usuario"], ETIQUETA_SIN_USUARIO, "Con usuario"))
dims_b = [("region", None), ("pais", None), ("categoria_de_equipo", None), ("clase_de_equipo", None),
          ("perfil_de_privilegio", None), ("estado_de_cuenta", None),
          ("banda_antiguedad", ORDEN_BANDA_ANT), ("banda_recencia", ORDEN_BANDA_REC), ("usuario", None)]
nrows = int(np.ceil(len(dims_b) / 2))
fig, axes = plt.subplots(nrows, 2, figsize=(15, 3.2 * nrows))
for ax, (dim, orden) in zip(axes.flat, dims_b):
    cifrado_por_b(vista_ab, dim, ax, orden)
for ax in axes.flat[len(dims_b):]:
    ax.axis("off")
handles, labels = axes.flat[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper right", ncol=3, frameon=False)
fig.suptitle("Estado de cifrado por dimensión de B (% de equipos)", x=0.01, ha="left", fontsize=13, fontweight="bold")
plt.tight_layout(rect=(0, 0, 1, 0.97))
plt.show()
```
