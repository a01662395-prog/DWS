```python

#Perfil del DF 
def perfil(df, cols):
    return pd.DataFrame({
        "tipo": df[cols].dtypes.astype(str),
        "nulos_%": (df[cols].isna().mean() * 100).round(1),
        "unicos": df[cols].nunique(),
        "top_5": [df[c].value_counts(dropna=False).head(5).to_dict() for c in cols],
    })

#Normalizar textos 
def norm_txt(s):
    """quita espacios, pasa a minúsculas y convierte los falsos nulos en NA"""
    s = s.astype("string").str.strip().str.lower()
    return s.replace({"": pd.NA, "n/a": pd.NA, "na": pd.NA, "null": pd.NA, "-": pd.NA, "none": pd.NA})

#Mapeo
def mapear(s, mapa, nombre):
    """aplica el mapeo y falla si aparece un valor no contemplado"""
    s = norm_txt(s)
    sin_mapear = set(s.dropna().unique()) - set(mapa)
    assert not sin_mapear, f"{nombre}: valores sin mapear {sin_mapear}"
    return s.map(mapa)
```
